```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

MODEL_NAME = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("Model loaded successfully!")
print("Using device:", device)
```

```python
def generate_sketch_idea(user_idea, style="Pencil Sketch", difficulty="Intermediate"):

    prompt = f"""
You are SketchMind, an AI assistant that helps artists design sketches.

Create a detailed sketch plan for this idea:

USER IDEA:
{user_idea}

STYLE:
{style}

DIFFICULTY:
{difficulty}

Return the result using exactly these sections:

TITLE:
MAIN SUBJECT:
COMPOSITION:
FOREGROUND:
MIDDLE GROUND:
BACKGROUND:
LIGHTING:
COLOR:
STYLE:
MOOD:
IMPORTANT DETAILS:
DRAWING STEPS:
1.
2.
3.
4.
5.
6.

Keep the answer practical for someone who wants to draw the sketch.
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=400,
        temperature=0.8,
        do_sample=True,
        top_p=0.9
    )

    result = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return result
```

```python
idea = "A lonely astronaut standing on an alien planet"

result = generate_sketch_idea(
    idea,
    style="Detailed Pencil Sketch",
    difficulty="Intermediate"
)

print(result)
```

```python
print("==========================================")
print("       SKETCHMIND AI")
print(" Transformer Sketch Idea Generator")
print("==========================================")

user_idea = input("\nEnter your sketch idea: ")

style = input(
    "Enter sketch style "
    "(Pencil / Charcoal / Cartoon / Realistic / Fantasy): "
)

difficulty = input(
    "Enter difficulty "
    "(Beginner / Intermediate / Advanced): "
)

print("\nGenerating your sketch concept...\n")

result = generate_sketch_idea(
    user_idea,
    style=style,
    difficulty=difficulty
)

print("==========================================")
print("             GENERATED DESIGN")
print("==========================================")
print(result)
```

```python
import gradio as gr

def sketchmind_app(idea, style, difficulty):
    if not idea.strip():
        return "Please enter a sketch idea."

    return generate_sketch_idea(
        idea,
        style=style,
        difficulty=difficulty
    )

interface = gr.Interface(
    fn=sketchmind_app,
    inputs=[
        gr.Textbox(
            label="Sketch Idea",
            placeholder="Example: A warrior standing in a magical forest"
        ),
        gr.Dropdown(
            [
                "Pencil Sketch",
                "Charcoal",
                "Realistic",
                "Cartoon",
                "Fantasy",
                "Concept Art"
            ],
            value="Pencil Sketch",
            label="Sketch Style"
        ),
        gr.Dropdown(
            [
                "Beginner",
                "Intermediate",
                "Advanced"
            ],
            value="Intermediate",
            label="Difficulty"
        )
    ],
    outputs=gr.Textbox(
        label="Generated Sketch Design",
        lines=25
    ),
    title="🎨 SketchMind - AI Sketch Idea Generator",
    description="Enter an idea and let the Transformer create a structured sketch plan."
)

interface.launch(share=True)
```
